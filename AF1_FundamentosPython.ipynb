# Ejercicio 1: Evaluador de Rendimiento de Modelos

Descripcion: Se define la función evaluar_modelo() para clasificar el rendimiento de un modelo según su exactitud (Accuracy). Con la función input() convertida a número decimal mediante float(), se pide al usuario un valor entre 0.0 y 1.0. Después se aplica un operador aritmético para obtener su equivalente en porcentaje y se utilizan condicionales (if, elif, else) junto con operadores lógicos (and, or) para ubicar el resultado en una de las cuatro categorías: Excelente (mayor o igual a 0.90), Aceptable (de 0.75 a menor de 0.90), Requiere Ajuste (de 0.50 a menor de 0.75) o Crítico / No apto (menor a 0.50), mostrando y retornando el diagnóstico final.

```python
def evaluar_modelo():
    # Solicitar al usuario el valor de exactitud entre 0.0 y 1.0
    exactitud = float(input("Ingresa la exactitud (Accuracy) del modelo (entre 0.0 y 1.0): "))

    # Operador aritmético para calcular el porcentaje equivalente
    porcentaje = exactitud * 100

    # Condicionales y operadores lógicos para determinar la categoría
    if exactitud < 0.0 or exactitud > 1.0:
        resultado = "Valor inválido (debe estar entre 0.0 y 1.0)"
    elif exactitud >= 0.90 and exactitud <= 1.0:
        resultado = "Excelente"
    elif exactitud >= 0.75 and exactitud < 0.90:
        resultado = "Aceptable"
    elif exactitud >= 0.50 and exactitud < 0.75:
        resultado = "Requiere Ajuste"
    else:
        resultado = "Crítico / No apto"

    print(f"Exactitud registrada: {exactitud} ({porcentaje:.1f}%) -> Categoría: {resultado}")
    return resultado

# Invocación de la función para probar la celda
evaluar_modelo()
```

# Ejercicio 2: Procesamiento de Hiperparámetros y Lotes
Descripcion: Se declara una tupla global inmutable llamada CONFIGURACION_BASE = (512, 10) para almacenar de forma fija las dimensiones de entrada y salida de una capa neuronal. Posteriormente se define la función procesar_tasas_aprendizaje(), la cual recibe una lista de tasas de aprendizaje (learning rates). Mediante un ciclo for, se recorre cada valor de la lista original, se multiplica por un factor de escala de 2.0 y se guarda en una nueva lista llamada tasas_escaladas. Finalmente, la función imprime en pantalla y retorna tanto la nueva lista modificada como la tupla base original.

```python
# 1. Tupla global inmutable con dimensiones fijas (entrada, salida)
CONFIGURACION_BASE = (512, 10)

# 2. Función que recibe una lista de tasas de aprendizaje
def procesar_tasas_aprendizaje(lista_tasas):
    tasas_escaladas = []
    factor_escala = 2.0

    # 3. Bucle for para multiplicar cada tasa por el factor de escala
    for tasa in lista_tasas:
        tasa_nueva = tasa * factor_escala
        tasas_escaladas.append(tasa_nueva)

    # 4. Imprimir y retornar la nueva lista y la tupla base
    print("Configuración base de la capa (Tupla fija):", CONFIGURACION_BASE)
    print("Tasas de aprendizaje originales:", lista_tasas)
    print("Tasas de aprendizaje escaladas (x2.0):", tasas_escaladas)

    return tasas_escaladas, CONFIGURACION_BASE

# Lista de prueba e invocación de la función
tasas_iniciales = [0.1, 0.01, 0.001, 0.0001]
procesar_tasas_aprendizaje(tasas_iniciales)
```

# Ejercicio 3: Simulador de Convergencia por Épocas
Se crea la función simular_entrenamiento() que pide al usuario mediante input() y la conversión int() el número máximo de épocas a simular. Se utiliza un bucle for desde la época 1 hasta el límite indicado por el usuario, calculando en cada iteración una pérdida (loss) decreciente con la operación matemática 1.0 dividido entre el número de época actual. Dentro del bucle, una estructura condicional if verifica en cada paso si la pérdida calculada es menor al umbral de tolerancia de 0.05; cuando esto se cumple, se imprime un mensaje avisando que el modelo convergió anticipadamente y se detiene el ciclo de inmediato usando la instrucción break.

```python
def simular_entrenamiento():
    # 1. Solicitar el número máximo de épocas como entero
    max_epocas = int(input("Ingresa el número máximo de épocas a simular: "))
    umbral = 0.05

    print(f"\nIniciando entrenamiento (Meta de pérdida < {umbral})...\n")

    # 2. Bucle for desde la época 1 hasta el límite indicado
    for epoca in range(1, max_epocas + 1):
        # 3. Operación matemática para simular una pérdida (loss) decreciente
        perdida = 1.0 / epoca
        print(f"Época {epoca} -> Pérdida (Loss): {perdida:.4f}")

        # 4. Condición if para detener el bucle si baja del umbral (0.05)
        if perdida < umbral:
            print(f"\n¡Convergencia anticipada alcanzada en la época {epoca}!")
            print(f"La pérdida ({perdida:.4f}) cayó por debajo del umbral ({umbral}). Deteniendo entrenamiento.")
            break
    else:
        print("\nSe alcanzó el límite de épocas sin llegar al umbral de convergencia.")

# Invocación de la función para probar la simulación
simular_entrenamiento()
```
