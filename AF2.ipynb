### Ejercicio 1.1.- Limpieza y Manipulación con Pandas

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
url = "https://archive.ics.uci.edu/ml/machine-learning-databases/wine-quality/winequality-red.csv"
df = pd.read_csv(url, sep=';')
```

```python
print("\nPrimeras 5 filas:")
display(df.head())
```

```python
print("\nResumen estadistico:")
display(df.describe())
```

```python
print(df.isnull().sum())
df = df.dropna()
```

```python
df["is_good"] = np.where(df["quality"] > 6, 1, 0)

print("\nColumna is_good:")
display(df.head())
```

```python
print("\nalcohol superior al 11%")
df_alcoholSuperior = df[df["alcohol"] > 11]
display(df_alcoholSuperior)
```

```python
print(len(df_alcoholSuperior))
```

#### Ejercicio 1.2.- Operaciones Matemáticas con NumPy

```python
x = df.drop(columns=["quality", "is_good"],errors="ignore").to_numpy()
```

```python
media = np.mean(x, axis=0)
desviacion = np.std(x, axis=0)
z = (x - media)/desviacion
print(z)
```

#####  Ejercicio 1.3.- Visualización de Insights con Matplotlib

```python
datos = np.array(df["alcohol"])

plt.hist(datos, bins=50, edgecolor='black')
plt.xlabel("Alcohol")
plt.ylabel("Frecuencia")
plt.title("Histograma de Alcohol")
plt.show()
```

```python
pH = np.array(df["pH"])
fixed_acidity = np.array(df["fixed acidity"])

plt.scatter(pH, fixed_acidity, color='red')
plt.xlabel("pH")
plt.ylabel("fixed acidity")
plt.title("Gráfico de dispersión")
plt.show()
```

##### Ejercicio 1.3  Visualización de Insights con Matplotlib

```python


# Agrupar los datos de alcohol por cada valor único de calidad
categorias_calidad = sorted(df["quality"].unique())
datos_boxplot = [df[df["quality"] == i]["alcohol"] for i in categorias_calidad]

# Crear el gráfico de caja y bigotes (Boxplot)
plt.figure(figsize=(8, 6))
plt.boxplot(datos_boxplot, tick_labels=categorias_calidad)

# Personalizar el gráfico
plt.title("Comparativa del Nivel de Alcohol por Calidad de Vino")
plt.xlabel("Calidad (Quality)")
plt.ylabel("Alcohol (%)")
plt.grid(axis='y', linestyle='--', alpha=0.5)

# Mostrar el gráfico
plt.show()
```

# **Ejercicio 2 .- Uso de Pandas, Matplotlib, Numpy**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

# Ejercicio 2.1.- Análisis Exploratorio con Pandas

```python
df = pd.read_csv('https://raw.githubusercontent.com/mwaskom/seaborn-data/master/penguins.csv')
```

```python
print("\nPrimeras 5 filas:")
display(df.head())
```

```python
print(df.info())
```

```python
display(df.isnull().sum())
```

```python
df_limpio = df.dropna(subset=[
    "body_mass_g",
    "bill_length_mm",
    "bill_depth_mm"
])
display(df_limpio.isnull().sum())
```

```python
promedio_especies = df.groupby('species')['body_mass_g'].mean()
display(promedio_especies)
```

```python
promedio_isla = df_limpio.groupby("island")["body_mass_g"].mean()

display(promedio_isla)
```

```python
gentoo_pesados = df_limpio[
    (df_limpio["species"] == "Gentoo") & (df_limpio["body_mass_g"] > 5000)]

display(gentoo_pesados)
```

# Ejercicio 2.2.- Procesamiento de Señales con NumPy

```python
matriz = picos = df_limpio[["bill_length_mm", "bill_depth_mm"]].to_numpy()
print(matriz)
```

```python
proporcion = picos[:, 0] / picos[:, 1]

print(proporcion)
```

```python
media = df_limpio.groupby("species")["bill_length_mm"].transform("mean")

desviacion = df_limpio.groupby("species")["bill_length_mm"].transform("std")

fuera_rango = df_limpio[
    np.abs(df_limpio["bill_length_mm"] - media) > (2 * desviacion)
]

display(fuera_rango)
```

```python
picos_pares = picos[::2]

print(picos_pares)
```

# Ejercicio 2.3.- Visualización Científica con Matplotlib

```python
plt.figure(figsize=(8, 5))

for especie in df_limpio["species"].unique():
    datos = df_limpio[df_limpio["species"] == especie]

    plt.scatter(
        datos["flipper_length_mm"],
        datos["body_mass_g"],
        label=especie
    )

plt.xlabel("Longitud de aleta (mm)")
plt.ylabel("Peso (g)")
plt.title("Peso vs. longitud de aleta")
plt.legend()
plt.show()
```

```python
indice = df_limpio["body_mass_g"].idxmax()

pesado = df_limpio.loc[indice]

plt.figure(figsize=(8, 5))

plt.scatter(
    df_limpio["flipper_length_mm"],
    df_limpio["body_mass_g"]
)

plt.scatter(
    pesado["flipper_length_mm"],
    pesado["body_mass_g"],
    color="red",
    s=100
)

plt.annotate(
    "Pingüino más pesado",
    (pesado["flipper_length_mm"], pesado["body_mass_g"])
)

plt.xlabel("Longitud de aleta (mm)")
plt.ylabel("Peso (g)")
plt.title("Identificación del pingüino más pesado")
plt.show()
```

```python
fig, ax = plt.subplots(1, 2, figsize=(12, 5))

# Primera gráfica: distribución del peso
ax[0].hist(df_limpio["body_mass_g"], bins=20)
ax[0].set_title("Distribución del peso")
ax[0].set_xlabel("Peso (g)")
ax[0].set_ylabel("Cantidad de pingüinos")

# Segunda gráfica: distribución de la longitud de aleta
ax[1].hist(df_limpio["flipper_length_mm"], bins=20)
ax[1].set_title("Distribución de longitud de aleta")
ax[1].set_xlabel("Longitud (mm)")
ax[1].set_ylabel("Cantidad de pingüinos")

plt.tight_layout()
plt.show()
```
